```python
# ---- Fixed phase definitions ----
phase_definitions = {
    "ascending": {
        "range": ("2010-05-13", "2011-12-31"),
        "years": [("2010-05-13", "2010-12-31"),
                  ("2011-01-01", "2011-12-31")]
    },
    "maximum": {
        "range": ("2012-01-01", "2014-12-31"),
        "years": [("2012-01-01", "2012-12-31"),
                  ("2013-01-01", "2013-12-31"),
                  ("2014-01-01", "2014-12-31")]
    },
    "declining": {
        "range": ("2015-01-01", "2018-12-31"),
        "years": [("2015-01-01", "2015-12-31"),
                  ("2016-01-01", "2016-12-31"),
                  ("2017-01-01", "2017-12-31"),
                  ("2018-01-01", "2018-12-31")]
    },
    "minimum": {
        "range": ("2019-01-01", "2020-12-31"),
        "years": [("2019-01-01", "2019-12-31"),
                  ("2020-01-01", "2020-12-31")]
    }
}


# ---- Utility: slice data by date range ----
def slice_by_dates(sw_dates, *arrays, start, end):
    mask = (sw_dates >= np.datetime64(start)) & (sw_dates <= np.datetime64(end))
    return (sw_dates[mask],) + tuple(arr[mask] for arr in arrays)


# ---- Utility: compute correlation safely ----
def compute_corr(x, y):
    mask = ~np.isnan(x) & ~np.isnan(y)
    if mask.sum() > 1:
        r, p = pearsonr(x[mask], y[mask])
        return r, p
    return np.nan, np.nan


def compute_rmse(obs, pred):
    mask = ~np.isnan(obs) & ~np.isnan(pred)
    if mask.sum() > 0:
        y_true = obs[mask]
        y_pred = pred[mask]
        mse = MeanSquaredError()(y_true, y_pred).numpy()
        rmse = float(np.sqrt(mse))
        
        n = len(y_true)
        # --- Uncertainty of MSE
        errors = y_true - y_pred
        sq_errors = errors ** 2
        mse_std = np.std(sq_errors, ddof=1) / np.sqrt(n)

        # --- Uncertainty of RMSE (propagation)
        rmse_std = (0.5 / rmse) * mse_std if rmse > 0 else np.nan
        return rmse, rmse_std
    return np.nan, np.nan


# ---- ICME vertical lines (black, short tick at top) ----
def add_icme_lines(ax, sw_dates, icme_dates):
    for i, icme_date in enumerate(icme_dates):
        if sw_dates.min() <= icme_date <= sw_dates.max():
            ax.axvline(x=icme_date, color="black", linestyle="-",
                       ymax=1, ymin=0.95, linewidth=2.5,
                       label="ICME Event" if i == 0 else "")


def add_hse_verticals(ax, sw_dates, y_obs, dat, color="brown"):
    _, time_final, _ = hsemetric.HSE(y_obs)
    for k, (start_idx, end_idx) in enumerate(time_final):
        if 0 <= start_idx < len(sw_dates) and 0 <= end_idx < len(sw_dates):
            ax.axvspan(sw_dates[start_idx], sw_dates[end_idx],
                       color=color, alpha=0.3,
                       label=f"{dat} HSE Event" if k == 0 else "")
    # ax.legend(loc="upper left")

# ---- Helper: split arrays at ICME dates ----
def split_at_icme(sw_dates, *arrays, icme_dates):
    """
    Split arrays at ICME dates -> returns list of segments.
    Each segment is (sw_seg, arr1_seg, arr2_seg, ...).
    """
    segments = []
    start_idx = 0
    icme_dates = sorted([d for d in icme_dates if sw_dates.min() <= d <= sw_dates.max()])

    for icme in icme_dates:
        cut_idx = np.searchsorted(sw_dates, icme)
        if cut_idx > start_idx:
            segs = (sw_dates[start_idx:cut_idx],) + tuple(arr[start_idx:cut_idx] for arr in arrays)
            segments.append(segs)
        start_idx = cut_idx
    # final segment
    if start_idx < len(sw_dates):
        segs = (sw_dates[start_idx:],) + tuple(arr[start_idx:] for arr in arrays)
        segments.append(segs)
    return segments


# ---- Main plotting function ----
def plot_phases(phases_data, icme_dates_dict, save_path, title_prefix="SpeedNet"):
    """
    Plot Ascending, Maximum, Declining, Minimum phases in one figure.
    - Preserves NaN gaps in data.
    - Breaks yearly lines at ICME dates (treated as gaps).
    - Adds ICME markers (black) and HSE start/end markers (green dashed).
    """

    fig = plt.figure(figsize=(34, 32))
    outer_gs = GridSpec(4, 3, figure=fig, width_ratios=[10, 10, 4], wspace=0.3, hspace=0.35)

    phase_order = ["ascending", "maximum", "declining", "minimum"]
    
    # generate labels (a) ... (l)
    panel_labels = [f"({letter})" for letter in string.ascii_lowercase[:12]]
    label_counter = 0

    for row, phase in enumerate(phase_order):
        sw_dates, y_obs, y_model, y_pers27, y_persN, delay = phases_data[phase]
        icme_dates = icme_dates_dict.get(phase, [])

        full_start, full_end = phase_definitions[phase]["range"]
        year_splits = phase_definitions[phase]["years"]

        # Slice full data for phase
        sw_dates_full, y_obs_full, y_model_full, y_pers27_full, y_persN_full = slice_by_dates(
            sw_dates, y_obs, y_model, y_pers27, y_persN,
            start=full_start, end=full_end
        )

        # --- Column 1: full phase plot ---
        ax_full = fig.add_subplot(outer_gs[row, 0])
        ax_full.plot(sw_dates_full, y_model_full, "o-", color="blue", markersize=3, label="SpeedNet-BM")
        ax_full.plot(sw_dates_full, y_pers27_full, "s-", color="red", markersize=3, label="27-day")
        ax_full.plot(sw_dates_full, y_persN_full, "^-", color="orange", markersize=3, label="N-day")
        ax_full.plot(sw_dates_full, y_obs_full, ".-", color="green", markersize=2, label="Observed")

        ax_full.set_title(f"{phase.capitalize()} Phase ({full_start} → {full_end})\nDelay={delay}",
                          fontsize=12, fontweight="bold")
        ax_full.set_ylabel("V (Km/s)", fontsize=10, fontweight="bold")
        ax_full.xaxis.set_major_formatter(mdates.DateFormatter("%Y-%m"))
        ax_full.tick_params(axis="x", rotation=45, labelsize=7)
        ax_full.legend(loc="upper left", fontsize=8)
        
        ax_full.text(0.01, 1.05, panel_labels[label_counter],
                     transform=ax_full.transAxes,
                     fontsize=14, fontweight="bold", va="top", ha="left")
        label_counter += 1

        # r + TS (SpeedNet-BM vs Obs)
        rmse_full, err_full = compute_rmse(y_obs_full, y_model_full)
        r_full, _ = compute_corr(y_obs_full, y_model_full)
        tp_full, fp_full, fn_full = hsemetric.BatchwiseHSE(y_obs_full, np.array(y_model_full))
        ts_full = tp_full / (tp_full + fp_full + fn_full) if (tp_full + fp_full + fn_full) > 0 else np.nan
        ax_full.text(0.98, 0.95, f"RMSE={rmse_full:.1f}±{err_full:.1f}, r={r_full:.2f}, TS={ts_full:.2f}",
                     transform=ax_full.transAxes, fontsize=9, fontweight="bold",
                     ha="right", va="top",
                     bbox=dict(boxstyle="round", facecolor="white", alpha=0.6))

        # --- Column 2: yearly splits ---
        inner_gs = GridSpecFromSubplotSpec(len(year_splits), 1,
                                        subplot_spec=outer_gs[row, 1], hspace=0.25)
        for i, (ystart, yend) in enumerate(year_splits):
            ax_y = fig.add_subplot(inner_gs[i, 0])
            sw_y, y_obs_y, y_model_y, y_pers27_y, y_persN_y = slice_by_dates(
                sw_dates, y_obs, y_model, y_pers27, y_persN,
                start=ystart, end=yend
            )

            
        
            # split into chunks by ICME dates
            year_icmes = [d for d in icme_dates if np.datetime64(ystart) <= d <= np.datetime64(yend)]
            segments = split_at_icme(sw_y, y_obs_y, y_model_y, y_pers27_y, y_persN_y,
                                    icme_dates=year_icmes)

            for j, (sw_seg, obs_seg, model_seg, p27_seg, pN_seg) in enumerate(segments):
                # Add labels only for very first subplot + first segment
                if row == 0 and i == 0 and j == 0:
                    ax_y.plot(sw_seg, np.ma.masked_invalid(model_seg), "o-", color="blue", markersize=3,
                            label="SpeedNet-BM")
                    ax_y.plot(sw_seg, np.ma.masked_invalid(obs_seg), ".-", color="green", markersize=2,
                            label="Observed")
                else:
                    ax_y.plot(sw_seg, np.ma.masked_invalid(model_seg), "o-", color="blue", markersize=3)
                    ax_y.plot(sw_seg, np.ma.masked_invalid(obs_seg), ".-", color="green", markersize=2)
            if i == 0:  # label only on top of the first yearly subplot
                ax_y.annotate(panel_labels[label_counter],
                              xy=(0.01, 1.02), xycoords="axes fraction",
                              fontsize=16, fontweight="bold",
                              ha="left", va="bottom")
        
            # ICME + HSE markers
            add_icme_lines(ax_y, sw_y, icme_dates)
            # add_hse_verticals(ax_y, sw_y, y_obs_y)
            add_hse_verticals(ax_y, sw_y, y_obs_y, dat=f'Observed', color="brown")
            add_hse_verticals(ax_y, sw_y, y_model_y, dat=f'SpeedNet-BM', color="blue")

            ax_y.set_title(f"{ystart} → {yend}", fontsize=9)
            ax_y.set_ylabel("V (Km/s)", fontsize=8, fontweight="bold")
            ax_y.xaxis.set_major_formatter(mdates.DateFormatter("%Y-%m"))
            ax_y.tick_params(axis="x", labelsize=6)

            # yearly r + TS + RMSE
            r, _ = compute_corr(y_obs_y, y_model_y)
            tp, fp, fn = hsemetric.BatchwiseHSE(y_obs_y, np.array(y_model_y))
            ts = tp / (tp + fp + fn) if (tp + fp + fn) > 0 else np.nan
            rmse, err = compute_rmse(y_obs_y, y_model_y)

            ax_y.text(0.98, 0.95,
                      f"r={r:.2f}, TS={ts:.2f}\nRMSE={rmse:.1f}±{err:.1f}",
                      transform=ax_y.transAxes, fontsize=7, fontweight="bold",
                      ha="right", va="top",
                      bbox=dict(boxstyle="round", facecolor="white", alpha=0.6))

            # Legend only once
            if row == 0 and i == 0:
                ax_y.legend(loc="upper left", fontsize=7)
                ax_y.set_title("Yearly Splits", fontsize=13, fontweight="bold", pad=30)
        label_counter += 1
       
        
        # --- Column 3: correlation scatter ---
        ax_corr = fig.add_subplot(outer_gs[row, 2])
        mask_m = ~np.isnan(y_obs_full) & ~np.isnan(y_model_full)
        mask_27 = ~np.isnan(y_obs_full) & ~np.isnan(y_pers27_full)
        mask_N = ~np.isnan(y_obs_full) & ~np.isnan(y_persN_full)

        ax_corr.scatter(y_obs_full[mask_m], y_model_full[mask_m],
                        alpha=0.6, s=12, c="blue", label="Obs vs SpeedNet-BM")
        ax_corr.scatter(y_obs_full[mask_27], y_pers27_full[mask_27],
                        alpha=0.6, s=12, c="red", label="Obs vs 27-day")
        ax_corr.scatter(y_obs_full[mask_N], y_persN_full[mask_N],
                        alpha=0.6, s=12, c="orange", label="Obs vs N-day")

        def add_fit_line(x, y, color, label_suffix):
            if len(x) > 1 and len(y) > 1:
                coeffs = np.polyfit(x, y, 1)
                poly = np.poly1d(coeffs)
                x_line = np.linspace(np.nanmin(x), np.nanmax(x), 100)
                ax_corr.plot(x_line, poly(x_line), color=color, linestyle="--", linewidth=1.2,
                             label=f"{label_suffix} fit: y={coeffs[0]:.2f}x+{coeffs[1]:.2f}")

        add_fit_line(y_obs_full[mask_m], y_model_full[mask_m], "blue", "SpeedNet-BM")
        add_fit_line(y_obs_full[mask_27], y_pers27_full[mask_27], "red", "27-day")
        add_fit_line(y_obs_full[mask_N], y_persN_full[mask_N], "orange", "N-day")

        r_model, p_model = compute_corr(y_obs_full, y_model_full)
        r_27, p_27 = compute_corr(y_obs_full, y_pers27_full)
        r_N, p_N = compute_corr(y_obs_full, y_persN_full)

        ax_corr.set_title("Correlation", fontsize=11, fontweight="bold")
        ax_corr.set_xlabel("Observed V (Km/s)", fontsize=9, fontweight="bold")
        ax_corr.set_ylabel("Predicted V (Km/s)", fontsize=9, fontweight="bold")
        ax_corr.legend(loc="upper left", fontsize=7)

        ax_corr.text(0.98, 0.02,
                     f"SpeedNet-BM: r={r_model:.2f}, p={p_model:.1e}\n"
                     f"27-day: r={r_27:.2f}, p={p_27:.1e}\n"
                     f"N-day: r={r_N:.2f}, p={p_N:.1e}",
                     transform=ax_corr.transAxes, fontsize=8,
                     va="bottom", ha="right",
                     bbox=dict(boxstyle="round", facecolor="white", alpha=0.7))

        
        ax_corr.text(0.01, 1.05, panel_labels[label_counter],
                     transform=ax_corr.transAxes,
                     fontsize=14, fontweight="bold", va="top", ha="left")
        label_counter += 1
        
        # X-axis label on last row
        if row == len(phase_order) - 1:
            ax_full.set_xlabel("Date", fontsize=10, fontweight="bold")
            ax_y.set_xlabel("Date", fontsize=10, fontweight="bold")
            for ax in fig.get_axes():
                if ax.get_subplotspec().colspan.start == 1:
                    ax.set_xlabel("Date", fontsize=10, fontweight="bold")

    # fig.suptitle(f"{title_prefix}: Solar Cycle Phases", fontsize=18, fontweight="bold", y=0.97)
    os.makedirs(os.path.dirname(save_path), exist_ok=True)
    base, ext = os.path.splitext(save_path)
    save_path_pdf = f"{base}.pdf"
    plt.savefig(save_path, dpi = 500, bbox_inches="tight")
    plt.savefig(save_path_pdf, dpi = 500, bbox_inches="tight")
    # plt.close()
    print(f"Saved full phase plot → {save_path}")
```
